```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
kan = pd.read_csv('../../../kan_feynman_results_KAN.csv')
mlpkan = pd.read_csv('../../../kan_feynman_results_MLPKAN.csv')
mlpkan_lbfgs = pd.read_csv('../../../kan_feynman_results_MLPKANLBFGS.csv')
mlpkan_optimized = pd.read_csv('../../../kan_feynman_results_MLPKANoptimized.csv')
mlpkan_optimized_lbfgs = pd.read_csv('../../../kan_feynman_results_MLPKANoptimizedLBFGS.csv')
```

```python
kan_times = kan[kan['R2 Score KAN'] > 0.99][['KAN time']]
mlpkan_times = mlpkan[mlpkan['R2 Score KAN'] > 0.99][['MLPKAN time']]
mlpkan_lbfgs_times = mlpkan_lbfgs[mlpkan_lbfgs['R2 Score KAN'] > 0.99][['MLPKAN time']]
mlpkan_optimized_times = mlpkan_optimized[mlpkan_optimized['R2 Score KAN'] > 0.99][['MLPKAN time']]
mlpkan_optimized_lbfgs_times = mlpkan_optimized_lbfgs[mlpkan_optimized_lbfgs['R2 Score KAN'] > 0.99][['MLPKAN time']]
print(len(kan_times), len(mlpkan_times), len(mlpkan_lbfgs_times), len(mlpkan_optimized_times), len(mlpkan_optimized_lbfgs_times))
```

```python
plt.figure(figsize=(10, 6))
plt.boxplot([kan_times['KAN time'], mlpkan_times['MLPKAN time'], mlpkan_lbfgs_times['MLPKAN time'], mlpkan_optimized_times['MLPKAN time'], mlpkan_optimized_lbfgs_times['MLPKAN time']],
            labels=['KAN', 'MLPKAN', 'MLPKAN LBFGS', 'MLPKAN Optimized', 'MLPKAN Optimized LBFGS'])
plt.ylabel('Time (seconds)')
plt.title('Time to achieve R2 Score > 0.99')
plt.grid(True, which="both", ls="--")
plt.show()
```
